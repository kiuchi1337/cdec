```python
import json
import shutil
from pathlib import Path

# Cleans all metadata from a Jupyter Notebook and creates a backup file (.bak).


def clean_metadata(
    input_path, output_path=None, create_backup=True, remove_attachments=False
):
  input_file = Path(input_path)

  # Default output path is the input path (overwriting the file)
  if output_path is None:
    output_file = input_file
  else:
    output_file = Path(output_path)

  # Create a backup file (.bak) before processing
  if create_backup:
    backup_file = input_file.with_suffix(input_file.suffix + ".bak")
    shutil.copy2(input_file, backup_file)

  with open(input_file, "r", encoding="utf-8") as f:
    nb = json.load(f)

  # Reset top-level notebook metadata
  nb["metadata"] = {}

  # Reset cell-level metadata and optional attachments
  for cell in nb.get("cells", []):
    cell["metadata"] = {}

    # Optionally remove inline markdown attachments (e.g. pasted Base64 images)
    if remove_attachments and "attachments" in cell:
      cell.pop("attachments", None)

  with open(output_file, "w", encoding="utf-8") as f:
    json.dump(nb, f, indent=1)


clean_metadata("notebook.ipynb") # Creates 'your_notebook.ipynb.bak' and cleans 'notebook.ipynb' in-place
```
