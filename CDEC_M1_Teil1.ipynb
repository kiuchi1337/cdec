Wenn ihr einen bestimmten Kartenausschnitt für eure Analysen haben wollt, könnt ihr dafür eine sogenannte *Bounding Box* erstellen. Das ist ein bestimmter Bereich auf der Karte, der durch Koordinaten eingegrenzt ist. Um also einen eigenen Bereich zu definieren, könnt ihr mithilfe des Polygon- oder Rectangle-Tools eine Fläche direkt auf der Karte einzeichnen. 

- ![Polygon-/Retangle-Tool](images/CDEC_M1_Teil1_bbox.png)

Polygon-/Retangle-Tool

:::{.callout-tip title=Tipp}

Wenn ihr unzufrieden mit eurem Bereich seid, könnt ihr über das Mülleimer-Symbol den Kasten löschen, auf "Clear All" klicken und anschließend einen neuen Kasten erstellen.

:::

Führt als nächstes den folgenden Code aus. Dieser speichert die eingezeichnete Bounding Box als Variable `aoi` ("area of interest") und exportiert den Bereich als *GeoJSON* (Datei). Wenn ihr oben keinen Bereich eingezeichnet habt, gibt die nachfolgende Zelle einen Error.

```python
if my_map.user_roi:
    # 1. Koordinaten extrahieren
    coords = my_map.user_roi['geometry']['coordinates'][0]
    lons, lats = zip(*coords)

    # 2. Bounding Box berechnen
    west, south, east, north = min(lons), min(lats), max(lons), max(lats)

    # 3. Variable 'aoi' im gewünschten Format speichern
    aoi = {
        "west": west,
        "south": south,
        "east": east,
        "north": north
    }

    # 4. Als GeoJSON Datei speichern
    with open('aoi.geojson', 'w') as f:
        json.dump(my_map.user_roi, f)

    print(f"✓ AOI als Dictionary im openEO Format gespeichert: {aoi}")
    print(f"✓ GeoJSON wurde in 'aoi.geojson' gespeichert.")

    # 5. Visuelle Bestätigung, hier kann der der Kasten angepasst werden
    map_confirmation = folium.Map()
    folium.Rectangle(
        bounds=[[south, west], [north, east]],
        color='blue',                                   # Farbe ändern (z.B. blue, green, red, ...)
        fill=True,                                      # Box-Füllung (True oder False)
        fill_opacity=0.1                                # Transparenz der Füllung ändern (Werte von 0 - 1, bei Dezimalzahlen immer mit einem Punkt (nicht Komma!) dazwischen angeben)
    ).add_to(map_confirmation)

    map_confirmation.fit_bounds([[south, west], [north, east]])
    display(map_confirmation)
else:
    print("⚠️ Bitte zuerst eine Fläche auf der Karte zeichnen!")
```

Nach Ausführung der oberen Zelle wird nochmal die Karte mit eurem gewählten Bereich gezeigt. Das ist nur um sicherzustellen, dass die Koordinaten korrekt in der Variable `aoi` gespeichert wurden.

# 4 Das Copernicus Dataspace Ecosystem (CDSE)

Für sämtliche eurer Analysen werden Datensätze benötigt. Dafür steht im CDSE ein großer Datenkatalog mit **Collections** zur Verfügung. Jede Collection enthält Datensätze, die ihr für eigene Projekte nutzen könnt. Doch nicht jede Collection eignet sich für jede Analyse. Das heißt, ihr müsst herausfiltern können, welche Collections die passenden Daten für eure Analysen haben und welche nicht. Das ist neben den vielen Programmcodes eine der schwierigsten Aufgaben bei eigenen Projekten!

:::{.callout-note}
#### Hinweis:

Hier kommt ihr zum [Datenkatalog](https://documentation.dataspace.copernicus.eu/APIs/openEO/Collections.html) auf der Copernicus-Website.

:::

Am Anfang einer Collection steht oft eine kurze Zusammenfassung zum Satelliten und weiter unten technische Details wie die Anzahl und Art der Bänder. Durch die Spektralbänder (auch Kanäle genannt) erfahrt ihr, welche Wellenlängenbereiche der Satellit abdeckt, also der Spektralbereich, in dem der Sensor des Satelliten arbeitet. Je schmaler die Bandbreite und je höher die Anzahl der Bänder ist, desto höher ist auch die spektrale Auflösung des Satelliten. Durch die Bänder erfahren wir also, welche Wellenlängen des sichtbaren und des nicht-sichtbaren Lichts (z.B. Infrarot oder Mikrowellen) des elektromagnetischen Spektrums abgedeckt werden. Das ist wichtig für die weiteren Analysen in diesem Modul.
